```python
# =============================================================================
# CELL 04 - BLOCK 3: ASSIGN BASE LGD USING PRODUCT TYPE
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(portfolio_lgd)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 07: LGD STRESS IMPACT")
print("BLOCK 3 : ASSIGN BASE LGD USING PRODUCT TYPE")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    base_lgd_mapping = {
        "Mortgage": 0.20,
        "Auto Loan": 0.35,
        "SME Loan": 0.40,
        "Corporate Loan": 0.45,
        "Personal Loan": 0.60,
        "Credit Card": 0.85,
    }
    fallback_base_lgd = 0.50

    portfolio_lgd["Base_LGD"] = (
        portfolio_lgd["Product_Type"]
        .map(base_lgd_mapping)
        .fillna(fallback_base_lgd)
        .astype("float64")
    )

    base_lgd_validation = pd.DataFrame({
        "Product_Type": [
            "Mortgage", "Auto Loan", "SME Loan", "Corporate Loan",
            "Personal Loan", "Credit Card", "Other / Unclassified"
        ],
        "Expected_Base_LGD": [0.20, 0.35, 0.40, 0.45, 0.60, 0.85, 0.50],
        "Configured_Base_LGD": [
            base_lgd_mapping["Mortgage"], base_lgd_mapping["Auto Loan"],
            base_lgd_mapping["SME Loan"], base_lgd_mapping["Corporate Loan"],
            base_lgd_mapping["Personal Loan"], base_lgd_mapping["Credit Card"],
            fallback_base_lgd
        ],
    })
    base_lgd_validation["Status"] = np.where(
        np.isclose(
            base_lgd_validation["Expected_Base_LGD"],
            base_lgd_validation["Configured_Base_LGD"]
        ),
        "PASS", "FAIL"
    )

    missing_base_lgd = int(portfolio_lgd["Base_LGD"].isna().sum())
    invalid_base_lgd = int(
        ((portfolio_lgd["Base_LGD"] < 0) | (portfolio_lgd["Base_LGD"] > 1)).sum()
    )
    unexpected_product_types = sorted(
        set(portfolio_lgd["Product_Type"].dropna().unique())
        - set(base_lgd_mapping.keys())
    )

    if missing_base_lgd > 0:
        raise ValueError("Missing Base_LGD values were created.")
    if invalid_base_lgd > 0:
        raise ValueError("Base_LGD contains values outside the valid 0-to-1 range.")

    output_record_count = len(portfolio_lgd)
    records_processed = output_record_count

    base_lgd_distribution = (
        portfolio_lgd[["Product_Type", "Base_LGD"]]
        .value_counts(dropna=False)
        .rename("Record_Count")
        .reset_index()
        .sort_values(by=["Base_LGD", "Product_Type"])
        .reset_index(drop=True)
    )

    print("=" * 80)
    print("BASE LGD ASSIGNMENT COMPLETED")
    print("=" * 80)
    display(base_lgd_validation)
    print("\nBase LGD Distribution:")
    display(base_lgd_distribution)
    print("Missing Base LGD Values:", missing_base_lgd)
    print("Invalid Base LGD Values:", invalid_base_lgd)
    print("Unexpected Product Types Using Fallback:", unexpected_product_types)
    print("Base LGD Validation Status: PASS")

    logger.info("Base LGD assignment completed successfully.")
    block_status = "SUCCESS"
except Exception:
    block_status = "FAILED"
    logger.exception("Block 3 failed during Base LGD assignment.")
    raise
finally:
    block_end_time = datetime.now()
    block_duration_seconds = round((block_end_time - block_start_time).total_seconds(), 3)
    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```

```python
# =============================================================================
# CELL 05 - BLOCK 4: ASSIGN LGD FACTOR USING SCENARIO
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(portfolio_lgd)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 07: LGD STRESS IMPACT")
print("BLOCK 4 : ASSIGN LGD FACTOR USING SCENARIO")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    lgd_factor_mapping = {
        "Baseline": 1.00,
        "Mild_Stress": 1.10,
        "Severe_Stress": 1.25,
    }
    fallback_lgd_factor = 1.00

    portfolio_lgd["LGD_Factor"] = (
        portfolio_lgd["Scenario_Name"]
        .map(lgd_factor_mapping)
        .fillna(fallback_lgd_factor)
        .astype("float64")
    )

    lgd_factor_validation = pd.DataFrame({
        "Scenario_Name": ["Baseline", "Mild_Stress", "Severe_Stress", "Other"],
        "Expected_LGD_Factor": [1.00, 1.10, 1.25, 1.00],
        "Configured_LGD_Factor": [1.00, 1.10, 1.25, fallback_lgd_factor],
    })
    lgd_factor_validation["Status"] = np.where(
        np.isclose(
            lgd_factor_validation["Expected_LGD_Factor"],
            lgd_factor_validation["Configured_LGD_Factor"]
        ),
        "PASS", "FAIL"
    )

    missing_lgd_factor = int(portfolio_lgd["LGD_Factor"].isna().sum())
    invalid_lgd_factor = int((portfolio_lgd["LGD_Factor"] <= 0).sum())
    unexpected_scenarios = sorted(
        set(portfolio_lgd["Scenario_Name"].dropna().unique())
        - set(lgd_factor_mapping.keys())
    )

    if missing_lgd_factor > 0:
        raise ValueError("Missing LGD_Factor values were created.")
    if invalid_lgd_factor > 0:
        raise ValueError("Invalid non-positive LGD factors were created.")
    if unexpected_scenarios:
        raise ValueError(f"Unexpected scenarios were found: {unexpected_scenarios}")

    output_record_count = len(portfolio_lgd)
    records_processed = output_record_count

    lgd_factor_distribution = (
        portfolio_lgd[["Scenario_Name", "LGD_Factor"]]
        .value_counts(dropna=False)
        .rename("Record_Count")
        .reset_index()
        .sort_values(by="LGD_Factor")
        .reset_index(drop=True)
    )

    print("=" * 80)
    print("LGD FACTOR ASSIGNMENT COMPLETED")
    print("=" * 80)
    display(lgd_factor_validation)
    print("\nScenario LGD Factor Distribution:")
    display(lgd_factor_distribution)
    print("Missing LGD Factors:", missing_lgd_factor)
    print("Invalid LGD Factors:", invalid_lgd_factor)
    print("Unexpected Scenarios:", unexpected_scenarios)
    print("LGD Factor Validation Status: PASS")

    logger.info("Scenario LGD Factor assignment completed successfully.")
    block_status = "SUCCESS"
except Exception:
    block_status = "FAILED"
    logger.exception("Block 4 failed during LGD Factor assignment.")
    raise
finally:
    block_end_time = datetime.now()
    block_duration_seconds = round((block_end_time - block_start_time).total_seconds(), 3)
    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```

```python
# =============================================================================
# CELL 06 - BLOCK 5: CALCULATE STRESSED LGD AND ASSIGN LGD BAND
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(portfolio_lgd)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 07: LGD STRESS IMPACT")
print("BLOCK 5 : CALCULATE STRESSED LGD AND LGD BAND")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    uncapped_stressed_lgd = portfolio_lgd["Base_LGD"] * portfolio_lgd["LGD_Factor"]
    portfolio_lgd["Stressed_LGD"] = uncapped_stressed_lgd.clip(upper=1.00)
    capped_lgd_records = int((uncapped_stressed_lgd > 1.00).sum())

    lgd_band_conditions = [
        portfolio_lgd["Stressed_LGD"] < 0.30,
        portfolio_lgd["Stressed_LGD"] <= 0.60,
    ]
    portfolio_lgd["LGD_Band"] = np.select(
        lgd_band_conditions,
        ["Low LGD", "Medium LGD"],
        default="High LGD"
    )

    missing_stressed_lgd = int(portfolio_lgd["Stressed_LGD"].isna().sum())
    invalid_stressed_lgd = int(
        ((portfolio_lgd["Stressed_LGD"] < 0) | (portfolio_lgd["Stressed_LGD"] > 1)).sum()
    )
    formula_validation_exceptions = int(
        (~np.isclose(
            portfolio_lgd["Stressed_LGD"],
            np.minimum(uncapped_stressed_lgd, 1.00)
        )).sum()
    )
    allowed_lgd_bands = ["Low LGD", "Medium LGD", "High LGD"]
    missing_lgd_band = int(portfolio_lgd["LGD_Band"].isna().sum())
    invalid_lgd_band = int((~portfolio_lgd["LGD_Band"].isin(allowed_lgd_bands)).sum())

    if missing_stressed_lgd > 0:
        raise ValueError("Missing Stressed_LGD values were created.")
    if invalid_stressed_lgd > 0:
        raise ValueError("Stressed_LGD contains values outside the 0-to-1 range.")
    if formula_validation_exceptions > 0:
        raise ValueError("Stressed_LGD formula or cap validation failed.")
    if missing_lgd_band > 0 or invalid_lgd_band > 0:
        raise ValueError("Invalid or missing LGD Band values were created.")

    output_record_count = len(portfolio_lgd)
    records_processed = output_record_count

    lgd_band_distribution = pd.crosstab(
        portfolio_lgd["Scenario_Name"],
        portfolio_lgd["LGD_Band"],
        margins=True,
        margins_name="Total"
    )

    print("=" * 80)
    print("STRESSED LGD AND LGD BAND CALCULATION COMPLETED")
    print("=" * 80)
    print("Missing Stressed LGD Values:", missing_stressed_lgd)
    print("Invalid Stressed LGD Values:", invalid_stressed_lgd)
    print("LGD Records Capped at 100%:", capped_lgd_records)
    print("Formula / Cap Validation Exceptions:", formula_validation_exceptions)
    print("Missing LGD Bands:", missing_lgd_band)
    print("Invalid LGD Bands:", invalid_lgd_band)
    print("Stressed LGD Formula Validation: PASS")
    print("\nScenario vs LGD Band Distribution:")
    display(lgd_band_distribution)
    print("\nSample LGD Stress Records:")
    display(portfolio_lgd[[
        "Customer_ID", "Loan_ID", "Product_Type", "Scenario_Name",
        "Base_LGD", "LGD_Factor", "Stressed_LGD", "LGD_Band"
    ]].head(25))

    logger.info("Stressed LGD and LGD Band calculations completed successfully.")
    block_status = "SUCCESS"
except Exception:
    block_status = "FAILED"
    logger.exception("Block 5 failed during Stressed LGD calculation.")
    raise
finally:
    block_end_time = datetime.now()
    block_duration_seconds = round((block_end_time - block_start_time).total_seconds(), 3)
    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```

```python
# =============================================================================
# CELL 07 - BLOCK 6: GENERATE SCENARIO-WISE LGD SUMMARY
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(portfolio_lgd)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 07: LGD STRESS IMPACT")
print("BLOCK 6 : GENERATE SCENARIO-WISE LGD SUMMARY")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    lgd_scenario_summary = (
        portfolio_lgd.groupby("Scenario_Name", sort=False)
        .agg(
            Record_Count=("Loan_ID", "size"),
            Base_LGD_Mean=("Base_LGD", "mean"),
            Base_LGD_Minimum=("Base_LGD", "min"),
            Base_LGD_Maximum=("Base_LGD", "max"),
            Stressed_LGD_Mean=("Stressed_LGD", "mean"),
            Stressed_LGD_Minimum=("Stressed_LGD", "min"),
            Stressed_LGD_Maximum=("Stressed_LGD", "max"),
        )
        .reset_index()
    )

    numeric_summary_columns = [
        "Base_LGD_Mean", "Base_LGD_Minimum", "Base_LGD_Maximum",
        "Stressed_LGD_Mean", "Stressed_LGD_Minimum", "Stressed_LGD_Maximum"
    ]
    lgd_scenario_summary[numeric_summary_columns] = (
        lgd_scenario_summary[numeric_summary_columns].round(4)
    )

    expected_scenario_counts = {
        "Baseline": 10000,
        "Mild_Stress": 10000,
        "Severe_Stress": 10000,
    }
    lgd_scenario_summary["Expected_Record_Count"] = (
        lgd_scenario_summary["Scenario_Name"].map(expected_scenario_counts)
    )
    lgd_scenario_summary["Status"] = np.where(
        lgd_scenario_summary["Record_Count"]
        == lgd_scenario_summary["Expected_Record_Count"],
        "PASS", "FAIL"
    )

    failed_scenario_summaries = lgd_scenario_summary.loc[
        lgd_scenario_summary["Status"] != "PASS"
    ]
    if not failed_scenario_summaries.empty:
        raise ValueError("Scenario-wise LGD summary validation failed.")

    output_record_count = len(lgd_scenario_summary)
    records_processed = input_record_count

    print("=" * 80)
    print("SCENARIO-WISE LGD SUMMARY")
    print("=" * 80)
    display(lgd_scenario_summary)
    print("Scenario-wise LGD Summary Status: PASS")

    logger.info("Scenario-wise LGD summary generated successfully.")
    block_status = "SUCCESS"
except Exception:
    block_status = "FAILED"
    logger.exception("Block 6 failed while generating LGD summary.")
    raise
finally:
    block_end_time = datetime.now()
    block_duration_seconds = round((block_end_time - block_start_time).total_seconds(), 3)
    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```

```python
# =============================================================================
# CELL 08 - BLOCK 7: FINAL VALIDATION AND RECONCILIATION
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(portfolio_pd)
output_record_count = len(portfolio_lgd)
records_processed = 0

print("=" * 80)
print("MODULE 07: LGD STRESS IMPACT")
print("BLOCK 7 : FINAL VALIDATION AND RECONCILIATION")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    final_duplicate_keys = int(
        portfolio_lgd[["Loan_ID", "Scenario_ID"]].duplicated().sum()
    )
    final_missing_base_lgd = int(portfolio_lgd["Base_LGD"].isna().sum())
    final_missing_lgd_factor = int(portfolio_lgd["LGD_Factor"].isna().sum())
    final_missing_stressed_lgd = int(portfolio_lgd["Stressed_LGD"].isna().sum())
    final_missing_lgd_band = int(portfolio_lgd["LGD_Band"].isna().sum())
    lgd_band_total = int(lgd_band_distribution.loc["Total", "Total"])

    validation_results = pd.DataFrame({
        "Validation_Check": [
            "Input and Output Record Count",
            "Expected Output Record Count",
            "Duplicate Loan-Scenario Keys",
            "Missing Base LGD",
            "Missing LGD Factor",
            "Missing Stressed LGD",
            "Missing LGD Band",
            "Invalid Base LGD",
            "Invalid Stressed LGD",
            "Formula / Cap Validation Exceptions",
            "LGD Band Distribution Total",
        ],
        "Expected_Result": [
            input_record_count, 30000, 0, 0, 0, 0, 0, 0, 0, 0, 30000
        ],
        "Actual_Result": [
            output_record_count, output_record_count, final_duplicate_keys,
            final_missing_base_lgd, final_missing_lgd_factor,
            final_missing_stressed_lgd, final_missing_lgd_band,
            invalid_base_lgd, invalid_stressed_lgd,
            formula_validation_exceptions, lgd_band_total
        ],
    })
    validation_results["Status"] = np.where(
        validation_results["Expected_Result"].astype(str)
        == validation_results["Actual_Result"].astype(str),
        "PASS", "FAIL"
    )

    lgd_stress_summary = pd.DataFrame({
        "Metric": [
            "Input Records", "Output Records", "Input Columns", "Output Columns",
            "Mortgage Base LGD", "Auto Loan Base LGD", "SME Loan Base LGD",
            "Corporate Loan Base LGD", "Personal Loan Base LGD",
            "Credit Card Base LGD", "Other Base LGD",
            "Baseline LGD Factor", "Mild Stress LGD Factor",
            "Severe Stress LGD Factor", "Minimum Stressed LGD",
            "Maximum Stressed LGD", "LGD Records Capped at 100%",
            "Low LGD Band Records", "Medium LGD Band Records",
            "High LGD Band Records", "Processing Timestamp"
        ],
        "Result": [
            input_record_count, output_record_count, input_column_count,
            len(portfolio_lgd.columns), 0.20, 0.35, 0.40, 0.45, 0.60, 0.85,
            0.50, 1.00, 1.10, 1.25,
            float(portfolio_lgd["Stressed_LGD"].min()),
            float(portfolio_lgd["Stressed_LGD"].max()), capped_lgd_records,
            int((portfolio_lgd["LGD_Band"] == "Low LGD").sum()),
            int((portfolio_lgd["LGD_Band"] == "Medium LGD").sum()),
            int((portfolio_lgd["LGD_Band"] == "High LGD").sum()),
            datetime.now()
        ],
    })

    failed_validations = validation_results.loc[
        validation_results["Status"] != "PASS"
    ]
    if not failed_validations.empty:
        raise ValueError("One or more Module 07 validation checks failed.")

    records_processed = output_record_count

    print("=" * 80)
    print("FINAL LGD STRESS IMPACT VALIDATION")
    print("=" * 80)
    display(validation_results)
    print("\nLGD Stress Module Summary:")
    display(lgd_stress_summary)
    print("Overall Module 07 Validation Status: PASS")

    logger.info("All Module 07 validation checks passed.")
    block_status = "SUCCESS"
except Exception:
    block_status = "FAILED"
    logger.exception("Block 7 failed during final validation.")
    raise
finally:
    block_end_time = datetime.now()
    block_duration_seconds = round((block_end_time - block_start_time).total_seconds(), 3)
    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```
