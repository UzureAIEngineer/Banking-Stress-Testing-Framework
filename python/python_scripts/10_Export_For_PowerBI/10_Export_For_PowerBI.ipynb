```python
# =============================================================================
# CELL 03 - BLOCK 2: LOAD AND VALIDATE MODULE 09 REPORTING DATASETS
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = 0
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 10: EXPORT FOR POWER BI")
print("BLOCK 2 : LOAD AND VALIDATE MODULE 09 REPORTING DATASETS")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    ecl_summary = pd.read_csv(
        ECL_SOURCE_FILE,
        low_memory=False
    )
    product_ecl_summary = pd.read_csv(
        PRODUCT_SOURCE_FILE,
        low_memory=False
    )
    segment_ecl_summary = pd.read_csv(
        SEGMENT_SOURCE_FILE,
        low_memory=False
    )

    ecl_required_columns = [
        "Scenario_Name",
        "Account_Count",
        "Total_Exposure",
        "Total_Base_ECL",
        "Total_Stressed_ECL",
        "Incremental_Stress_Loss",
    ]
    product_required_columns = [
        "Scenario_Name",
        "Product_Type",
        "Account_Count",
        "Total_Exposure",
        "Total_Stressed_ECL",
    ]
    segment_required_columns = [
        "Scenario_Name",
        "Risk_Segment",
        "Account_Count",
        "Total_Exposure",
        "Total_Stressed_ECL",
    ]

    missing_ecl_columns = sorted(
        set(ecl_required_columns) - set(ecl_summary.columns)
    )
    missing_product_columns = sorted(
        set(product_required_columns) - set(product_ecl_summary.columns)
    )
    missing_segment_columns = sorted(
        set(segment_required_columns) - set(segment_ecl_summary.columns)
    )

    if missing_ecl_columns:
        raise ValueError(
            f"Executive summary columns are missing: {missing_ecl_columns}"
        )
    if missing_product_columns:
        raise ValueError(
            f"Product summary columns are missing: {missing_product_columns}"
        )
    if missing_segment_columns:
        raise ValueError(
            f"Segment summary columns are missing: {missing_segment_columns}"
        )

    ecl_missing_values = int(
        ecl_summary[ecl_required_columns].isna().sum().sum()
    )
    product_missing_values = int(
        product_ecl_summary[product_required_columns].isna().sum().sum()
    )
    segment_missing_values = int(
        segment_ecl_summary[segment_required_columns].isna().sum().sum()
    )

    ecl_duplicate_keys = int(
        ecl_summary[["Scenario_Name"]].duplicated().sum()
    )
    product_duplicate_keys = int(
        product_ecl_summary[
            ["Scenario_Name", "Product_Type"]
        ].duplicated().sum()
    )
    segment_duplicate_keys = int(
        segment_ecl_summary[
            ["Scenario_Name", "Risk_Segment"]
        ].duplicated().sum()
    )

    expected_scenarios = {
        "Baseline",
        "Mild_Stress",
        "Severe_Stress",
    }
    ecl_scenarios = set(ecl_summary["Scenario_Name"].unique())
    product_scenarios = set(product_ecl_summary["Scenario_Name"].unique())
    segment_scenarios = set(segment_ecl_summary["Scenario_Name"].unique())

    ecl_account_total = int(ecl_summary["Account_Count"].sum())
    product_account_total = int(
        product_ecl_summary["Account_Count"].sum()
    )
    segment_account_total = int(
        segment_ecl_summary["Account_Count"].sum()
    )

    if len(ecl_summary) != 3:
        raise ValueError(
            "Executive summary does not contain three scenario rows."
        )
    if ecl_missing_values + product_missing_values + segment_missing_values > 0:
        raise ValueError(
            "Missing values were found in required Power BI fields."
        )
    if ecl_duplicate_keys + product_duplicate_keys + segment_duplicate_keys > 0:
        raise ValueError(
            "Duplicate reporting keys were found."
        )
    if not (
        ecl_scenarios
        == product_scenarios
        == segment_scenarios
        == expected_scenarios
    ):
        raise ValueError(
            "Scenario values are inconsistent across reporting datasets."
        )
    if not (
        ecl_account_total
        == product_account_total
        == segment_account_total
        == 30000
    ):
        raise ValueError(
            "Account totals do not reconcile across reporting datasets."
        )

    input_record_count = (
        len(ecl_summary)
        + len(product_ecl_summary)
        + len(segment_ecl_summary)
    )
    records_processed = input_record_count

    print("=" * 80)
    print("MODULE 09 REPORTING DATASETS LOADED SUCCESSFULLY")
    print("=" * 80)
    print("Executive Summary Rows:", len(ecl_summary))
    print("Product Summary Rows:", len(product_ecl_summary))
    print("Segment Summary Rows:", len(segment_ecl_summary))
    print("Combined Input Rows:", input_record_count)
    print("Executive Account Total:", ecl_account_total)
    print("Product Account Total:", product_account_total)
    print("Segment Account Total:", segment_account_total)
    print("Executive Missing Values:", ecl_missing_values)
    print("Product Missing Values:", product_missing_values)
    print("Segment Missing Values:", segment_missing_values)
    print("Executive Duplicate Keys:", ecl_duplicate_keys)
    print("Product Duplicate Keys:", product_duplicate_keys)
    print("Segment Duplicate Keys:", segment_duplicate_keys)
    print("Input Validation Status: PASS")

    print("\nExecutive Summary:")
    display(ecl_summary.round(2))

    logger.info(
        "Module 09 reporting datasets loaded and validated successfully."
    )
    block_status = "SUCCESS"

except Exception:
    block_status = "FAILED"
    logger.exception(
        "Block 2 failed while loading Module 09 reporting datasets."
    )
    raise

finally:
    block_end_time = datetime.now()
    block_duration_seconds = round(
        (block_end_time - block_start_time).total_seconds(),
        3
    )
    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```

```python
# =============================================================================
# CELL 04 - BLOCK 3: PREPARE POWER BI EXPORT DATASETS
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = (
    len(ecl_summary)
    + len(product_ecl_summary)
    + len(segment_ecl_summary)
)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 10: EXPORT FOR POWER BI")
print("BLOCK 3 : PREPARE POWER BI EXPORT DATASETS")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    # Preserve the exact SAS V1 reporting structures and column names.
    powerbi_ecl_summary = ecl_summary[
        ecl_required_columns
    ].copy(deep=True)

    powerbi_product_summary = product_ecl_summary[
        product_required_columns
    ].copy(deep=True)

    powerbi_segment_summary = segment_ecl_summary[
        segment_required_columns
    ].copy(deep=True)

    # Apply stable business ordering for consistent Power BI refreshes.
    scenario_order = {
        "Baseline": 1,
        "Mild_Stress": 2,
        "Severe_Stress": 3,
    }

    powerbi_ecl_summary["_Scenario_Order"] = (
        powerbi_ecl_summary["Scenario_Name"].map(scenario_order)
    )
    powerbi_ecl_summary = (
        powerbi_ecl_summary
        .sort_values("_Scenario_Order", kind="stable")
        .drop(columns="_Scenario_Order")
        .reset_index(drop=True)
    )

    powerbi_product_summary["_Scenario_Order"] = (
        powerbi_product_summary["Scenario_Name"].map(scenario_order)
    )
    powerbi_product_summary = (
        powerbi_product_summary
        .sort_values(
            ["_Scenario_Order", "Product_Type"],
            kind="stable"
        )
        .drop(columns="_Scenario_Order")
        .reset_index(drop=True)
    )

    powerbi_segment_summary["_Scenario_Order"] = (
        powerbi_segment_summary["Scenario_Name"].map(scenario_order)
    )
    powerbi_segment_summary = (
        powerbi_segment_summary
        .sort_values(
            ["_Scenario_Order", "Risk_Segment"],
            kind="stable"
        )
        .drop(columns="_Scenario_Order")
        .reset_index(drop=True)
    )

    prepared_missing_values = int(
        powerbi_ecl_summary.isna().sum().sum()
        + powerbi_product_summary.isna().sum().sum()
        + powerbi_segment_summary.isna().sum().sum()
    )

    structure_exceptions = 0
    structure_exceptions += int(
        list(powerbi_ecl_summary.columns) != ecl_required_columns
    )
    structure_exceptions += int(
        list(powerbi_product_summary.columns)
        != product_required_columns
    )
    structure_exceptions += int(
        list(powerbi_segment_summary.columns)
        != segment_required_columns
    )
    structure_exceptions += int(
        len(powerbi_ecl_summary) != len(ecl_summary)
    )
    structure_exceptions += int(
        len(powerbi_product_summary) != len(product_ecl_summary)
    )
    structure_exceptions += int(
        len(powerbi_segment_summary) != len(segment_ecl_summary)
    )

    if prepared_missing_values > 0:
        raise ValueError(
            "Prepared Power BI datasets contain missing values."
        )
    if structure_exceptions > 0:
        raise ValueError(
            "Prepared Power BI dataset structure validation failed."
        )

    output_record_count = (
        len(powerbi_ecl_summary)
        + len(powerbi_product_summary)
        + len(powerbi_segment_summary)
    )
    records_processed = input_record_count

    print("=" * 80)
    print("POWER BI DATASETS PREPARED")
    print("=" * 80)
    print("Executive Export Rows:", len(powerbi_ecl_summary))
    print("Product Export Rows:", len(powerbi_product_summary))
    print("Segment Export Rows:", len(powerbi_segment_summary))
    print("Combined Export Rows:", output_record_count)
    print("Prepared Missing Values:", prepared_missing_values)
    print("Structure Exceptions:", structure_exceptions)
    print("Power BI Preparation Status: PASS")

    print("\nProduct Summary Sample:")
    display(powerbi_product_summary.head(12).round(2))

    print("\nSegment Summary Sample:")
    display(powerbi_segment_summary.head(9).round(2))

    logger.info(
        "Power BI export datasets prepared successfully."
    )
    block_status = "SUCCESS"

except Exception:
    block_status = "FAILED"
    logger.exception(
        "Block 3 failed while preparing Power BI datasets."
    )
    raise

finally:
    block_end_time = datetime.now()
    block_duration_seconds = round(
        (block_end_time - block_start_time).total_seconds(),
        3
    )
    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)

```

```python

# =============================================================================
# CELL 05 - BLOCK 4: EXPORT CSV FILES FOR POWER BI
# SAS V1: PROC EXPORT DBMS=CSV REPLACE
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = output_record_count
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 10: EXPORT FOR POWER BI")
print("BLOCK 4 : EXPORT CSV FILES FOR POWER BI")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    powerbi_ecl_summary.to_csv(
        ECL_EXPORT_FILE,
        index=False,
        encoding="utf-8"
    )
    powerbi_product_summary.to_csv(
        PRODUCT_EXPORT_FILE,
        index=False,
        encoding="utf-8"
    )
    powerbi_segment_summary.to_csv(
        SEGMENT_EXPORT_FILE,
        index=False,
        encoding="utf-8"
    )

    exported_files = [
        ECL_EXPORT_FILE,
        PRODUCT_EXPORT_FILE,
        SEGMENT_EXPORT_FILE,
    ]

    output_record_count = len(exported_files)
    records_processed = input_record_count

    print("=" * 80)
    print("POWER BI CSV EXPORT COMPLETED")
    print("=" * 80)
    print("1.", ECL_EXPORT_FILE)
    print("2.", PRODUCT_EXPORT_FILE)
    print("3.", SEGMENT_EXPORT_FILE)
    print("Exported File Count:", output_record_count)

    logger.info(
        "Three Power BI CSV files exported successfully."
    )
    block_status = "SUCCESS"

except Exception:
    block_status = "FAILED"
    logger.exception(
        "Block 4 failed while exporting Power BI CSV files."
    )
    raise

finally:
    block_end_time = datetime.now()
    block_duration_seconds = round(
        (block_end_time - block_start_time).total_seconds(),
        3
    )
    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT FILES       :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


# =============================================================================
# CELL 06 - BLOCK 5: VERIFY EXPORTED FILES AND DATA CONTENT
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(exported_files)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 10: EXPORT FOR POWER BI")
print("BLOCK 5 : VERIFY EXPORTED FILES AND DATA CONTENT")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    saved_ecl_summary = pd.read_csv(
        ECL_EXPORT_FILE,
        low_memory=False
    )
    saved_product_summary = pd.read_csv(
        PRODUCT_EXPORT_FILE,
        low_memory=False
    )
    saved_segment_summary = pd.read_csv(
        SEGMENT_EXPORT_FILE,
        low_memory=False
    )

    def sha256_file(file_path):
        digest = hashlib.sha256()
        with open(file_path, "rb") as file_handle:
            for chunk in iter(lambda: file_handle.read(65536), b""):
                digest.update(chunk)
        return digest.hexdigest()

    export_verification = pd.DataFrame({
        "Dataset_Name": [
            "Executive ECL Summary",
            "Product ECL Summary",
            "Segment ECL Summary",
        ],
        "File_Name": [
            ECL_EXPORT_FILE.name,
            PRODUCT_EXPORT_FILE.name,
            SEGMENT_EXPORT_FILE.name,
        ],
        "Exists": [
            ECL_EXPORT_FILE.exists(),
            PRODUCT_EXPORT_FILE.exists(),
            SEGMENT_EXPORT_FILE.exists(),
        ],
        "Size_Bytes": [
            ECL_EXPORT_FILE.stat().st_size,
            PRODUCT_EXPORT_FILE.stat().st_size,
            SEGMENT_EXPORT_FILE.stat().st_size,
        ],
        "Expected_Rows": [
            len(powerbi_ecl_summary),
            len(powerbi_product_summary),
            len(powerbi_segment_summary),
        ],
        "Saved_Rows": [
            len(saved_ecl_summary),
            len(saved_product_summary),
            len(saved_segment_summary),
        ],
        "Expected_Columns": [
            len(powerbi_ecl_summary.columns),
            len(powerbi_product_summary.columns),
            len(powerbi_segment_summary.columns),
        ],
        "Saved_Columns": [
            len(saved_ecl_summary.columns),
            len(saved_product_summary.columns),
            len(saved_segment_summary.columns),
        ],
        "SHA256": [
            sha256_file(ECL_EXPORT_FILE),
            sha256_file(PRODUCT_EXPORT_FILE),
            sha256_file(SEGMENT_EXPORT_FILE),
        ],
        "Full_Path": [
            str(ECL_EXPORT_FILE),
            str(PRODUCT_EXPORT_FILE),
            str(SEGMENT_EXPORT_FILE),
        ],
    })

    export_verification["Status"] = np.where(
        export_verification["Exists"]
        & (export_verification["Size_Bytes"] > 0)
        & (
            export_verification["Expected_Rows"]
            == export_verification["Saved_Rows"]
        )
        & (
            export_verification["Expected_Columns"]
            == export_verification["Saved_Columns"]
        ),
        "PASS",
        "FAIL"
    )

    numerical_reconciliation_exceptions = 0
    numerical_reconciliation_exceptions += int(
        not np.allclose(
            saved_ecl_summary[
                [
                    "Account_Count",
                    "Total_Exposure",
                    "Total_Base_ECL",
                    "Total_Stressed_ECL",
                    "Incremental_Stress_Loss",
                ]
            ].to_numpy(dtype="float64"),
            powerbi_ecl_summary[
                [
                    "Account_Count",
                    "Total_Exposure",
                    "Total_Base_ECL",
                    "Total_Stressed_ECL",
                    "Incremental_Stress_Loss",
                ]
            ].to_numpy(dtype="float64")
        )
    )
    numerical_reconciliation_exceptions += int(
        not np.allclose(
            saved_product_summary[
                [
                    "Account_Count",
                    "Total_Exposure",
                    "Total_Stressed_ECL",
                ]
            ].to_numpy(dtype="float64"),
            powerbi_product_summary[
                [
                    "Account_Count",
                    "Total_Exposure",
                    "Total_Stressed_ECL",
                ]
            ].to_numpy(dtype="float64")
        )
    )
    numerical_reconciliation_exceptions += int(
        not np.allclose(
            saved_segment_summary[
                [
                    "Account_Count",
                    "Total_Exposure",
                    "Total_Stressed_ECL",
                ]
            ].to_numpy(dtype="float64"),
            powerbi_segment_summary[
                [
                    "Account_Count",
                    "Total_Exposure",
                    "Total_Stressed_ECL",
                ]
            ].to_numpy(dtype="float64")
        )
    )

    if not (export_verification["Status"] == "PASS").all():
        raise ValueError(
            "One or more Power BI export files failed verification."
        )
    if numerical_reconciliation_exceptions > 0:
        raise ValueError(
            "Exported numeric values do not reconcile to source datasets."
        )

    output_record_count = len(export_verification)
    records_processed = sum(
        export_verification["Saved_Rows"]
    )

    print("=" * 80)
    print("POWER BI EXPORT VERIFICATION")
    print("=" * 80)
    display(export_verification)
    print(
        "Numerical Reconciliation Exceptions:",
        numerical_reconciliation_exceptions
    )
    print("Power BI Export Verification Status: PASS")

    logger.info(
        "Power BI exported files verified successfully."
    )
    block_status = "SUCCESS"

except Exception:
    block_status = "FAILED"
    logger.exception(
        "Block 5 failed during exported-file verification."
    )
    raise

finally:
    block_end_time = datetime.now()
    block_duration_seconds = round(
        (block_end_time - block_start_time).total_seconds(),
        3
    )
    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT FILES        :", input_record_count)
    print("VERIFIED FILES     :", output_record_count)
    print("RECORDS VERIFIED   :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```

```python
# =============================================================================
# CELL 07 - BLOCK 6: CREATE VALIDATION EVIDENCE AND EXPORT MANIFEST
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(export_verification)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 10: EXPORT FOR POWER BI")
print("BLOCK 6 : CREATE VALIDATION EVIDENCE AND EXPORT MANIFEST")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    validation_results = pd.DataFrame({
        "Validation_Check": [
            "Executive Source File Exists",
            "Product Source File Exists",
            "Segment Source File Exists",
            "Executive Summary Row Count",
            "Executive Summary Account Total",
            "Product Summary Account Total",
            "Segment Summary Account Total",
            "Executive Missing Required Values",
            "Product Missing Required Values",
            "Segment Missing Required Values",
            "Executive Duplicate Keys",
            "Product Duplicate Keys",
            "Segment Duplicate Keys",
            "Expected Export File Count",
            "Verified Export File Count",
            "Failed Export File Count",
            "Numerical Reconciliation Exceptions",
        ],
        "Expected_Result": [
            True,
            True,
            True,
            3,
            30000,
            30000,
            30000,
            0,
            0,
            0,
            0,
            0,
            0,
            3,
            3,
            0,
            0,
        ],
        "Actual_Result": [
            ECL_SOURCE_FILE.exists(),
            PRODUCT_SOURCE_FILE.exists(),
            SEGMENT_SOURCE_FILE.exists(),
            len(ecl_summary),
            ecl_account_total,
            product_account_total,
            segment_account_total,
            ecl_missing_values,
            product_missing_values,
            segment_missing_values,
            ecl_duplicate_keys,
            product_duplicate_keys,
            segment_duplicate_keys,
            len(exported_files),
            int((export_verification["Status"] == "PASS").sum()),
            int((export_verification["Status"] != "PASS").sum()),
            numerical_reconciliation_exceptions,
        ],
    })

    validation_results["Status"] = np.where(
        validation_results["Expected_Result"].astype(str)
        == validation_results["Actual_Result"].astype(str),
        "PASS",
        "FAIL"
    )

    failed_validations = validation_results.loc[
        validation_results["Status"] != "PASS"
    ]
    if not failed_validations.empty:
        raise ValueError(
            "One or more Module 10 validation checks failed."
        )

    export_manifest = export_verification.copy(deep=True)
    export_manifest["Export_Timestamp"] = datetime.now().strftime(
        "%Y-%m-%d %H:%M:%S"
    )
    export_manifest["Source_Module"] = "09_Portfolio_Stress_Reporting"
    export_manifest["Target_System"] = "Power BI"
    export_manifest["Encoding"] = "UTF-8"
    export_manifest["Delimiter"] = "Comma"
    export_manifest["Header_Row"] = True

    export_sample = pd.concat(
        [
            powerbi_ecl_summary.assign(
                Export_Dataset="Executive ECL Summary"
            ),
            powerbi_product_summary.head(5).assign(
                Export_Dataset="Product ECL Summary"
            ),
            powerbi_segment_summary.head(5).assign(
                Export_Dataset="Segment ECL Summary"
            ),
        ],
        ignore_index=True,
        sort=False
    )

    validation_results.to_csv(
        EXPORT_VALIDATION_FILE,
        index=False,
        encoding="utf-8"
    )
    export_manifest.to_csv(
        EXPORT_MANIFEST_FILE,
        index=False,
        encoding="utf-8"
    )
    export_sample.to_csv(
        EXPORT_SAMPLE_FILE,
        index=False,
        encoding="utf-8"
    )

    evidence_files = [
        EXPORT_VALIDATION_FILE,
        EXPORT_MANIFEST_FILE,
        EXPORT_SAMPLE_FILE,
    ]
    if not all(file_path.exists() for file_path in evidence_files):
        raise FileNotFoundError(
            "One or more Module 10 evidence files were not created."
        )
    if not all(file_path.stat().st_size > 0 for file_path in evidence_files):
        raise ValueError(
            "One or more Module 10 evidence files are empty."
        )

    output_record_count = len(evidence_files)
    records_processed = len(export_verification)

    print("=" * 80)
    print("FINAL MODULE 10 VALIDATION")
    print("=" * 80)
    display(validation_results)

    print("\nPower BI Export Manifest:")
    display(export_manifest)

    print("Overall Module 10 Validation Status: PASS")

    logger.info(
        "Module 10 validation evidence and export manifest created successfully."
    )
    block_status = "SUCCESS"

except Exception:
    block_status = "FAILED"
    logger.exception(
        "Block 6 failed while creating validation evidence."
    )
    raise

finally:
    block_end_time = datetime.now()
    block_duration_seconds = round(
        (block_end_time - block_start_time).total_seconds(),
        3
    )
    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT FILES        :", input_record_count)
    print("EVIDENCE FILES     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)

```
