```python
# =============================================================================
# CELL 04 - BLOCK 3: ASSIGN BASE PD USING RISK SEGMENT
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(portfolio_pd)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 06: PD STRESS IMPACT")
print("BLOCK 3 : ASSIGN BASE PD USING RISK SEGMENT")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    base_pd_mapping = {
        "Low Risk": 0.02,
        "Medium Risk": 0.05,
        "High Risk": 0.10,
    }

    portfolio_pd["Base_PD"] = (
        portfolio_pd["Risk_Segment"]
        .map(base_pd_mapping)
        .fillna(0.15)
        .astype("float64")
    )

    base_pd_validation = pd.DataFrame({
        "Risk_Segment": [
            "Low Risk",
            "Medium Risk",
            "High Risk",
            "Other / Unclassified",
        ],
        "Expected_Base_PD": [0.02, 0.05, 0.10, 0.15],
        "Configured_Base_PD": [0.02, 0.05, 0.10, 0.15],
    })

    base_pd_validation["Status"] = np.where(
        np.isclose(
            base_pd_validation["Expected_Base_PD"],
            base_pd_validation["Configured_Base_PD"],
        ),
        "PASS",
        "FAIL",
    )

    missing_base_pd = int(portfolio_pd["Base_PD"].isna().sum())

    invalid_base_pd = int(
        ((portfolio_pd["Base_PD"] <= 0) | (portfolio_pd["Base_PD"] > 1)).sum()
    )

    if missing_base_pd > 0:
        raise ValueError("Missing Base_PD values were created.")

    if invalid_base_pd > 0:
        raise ValueError(
            "Base_PD contains values outside the valid 0-to-1 range."
        )

    output_record_count = len(portfolio_pd)
    records_processed = output_record_count

    base_pd_distribution = (
        portfolio_pd[["Risk_Segment", "Base_PD"]]
        .value_counts(dropna=False)
        .rename("Record_Count")
        .reset_index()
        .sort_values(by=["Base_PD", "Risk_Segment"])
        .reset_index(drop=True)
    )

    print("=" * 80)
    print("BASE PD ASSIGNMENT COMPLETED")
    print("=" * 80)
    display(base_pd_validation)
    print("\nBase PD Distribution:")
    display(base_pd_distribution)
    print("Missing Base PD Values:", missing_base_pd)
    print("Invalid Base PD Values:", invalid_base_pd)
    print("Base PD Validation Status: PASS")

    logger.info("Base PD assignment completed successfully.")
    block_status = "SUCCESS"

except Exception:
    block_status = "FAILED"
    logger.exception("Block 3 failed during Base PD assignment.")
    raise

finally:
    block_end_time = datetime.now()
    block_duration_seconds = round(
        (block_end_time - block_start_time).total_seconds(), 3
    )

    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```

```python
# =============================================================================
# CELL 05 - BLOCK 4: ASSIGN PD FACTOR USING SCENARIO
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(portfolio_pd)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 06: PD STRESS IMPACT")
print("BLOCK 4 : ASSIGN PD FACTOR USING SCENARIO")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    pd_factor_mapping = {
        "Baseline": 1.00,
        "Mild_Stress": 1.25,
        "Severe_Stress": 1.75,
    }

    portfolio_pd["PD_Factor"] = (
        portfolio_pd["Scenario_Name"]
        .map(pd_factor_mapping)
        .fillna(1.00)
        .astype("float64")
    )

    pd_factor_validation = pd.DataFrame({
        "Scenario_Name": ["Baseline", "Mild_Stress", "Severe_Stress"],
        "Expected_PD_Factor": [1.00, 1.25, 1.75],
        "Configured_PD_Factor": [1.00, 1.25, 1.75],
    })

    pd_factor_validation["Status"] = np.where(
        np.isclose(
            pd_factor_validation["Expected_PD_Factor"],
            pd_factor_validation["Configured_PD_Factor"],
        ),
        "PASS",
        "FAIL",
    )

    missing_pd_factor = int(portfolio_pd["PD_Factor"].isna().sum())
    invalid_pd_factor = int((portfolio_pd["PD_Factor"] <= 0).sum())

    unexpected_scenarios = sorted(
        set(portfolio_pd["Scenario_Name"].dropna().unique())
        - set(pd_factor_mapping.keys())
    )

    if missing_pd_factor > 0:
        raise ValueError("Missing PD_Factor values were created.")

    if invalid_pd_factor > 0:
        raise ValueError("Invalid non-positive PD factors were created.")

    if unexpected_scenarios:
        raise ValueError(
            f"Unexpected scenarios were found: {unexpected_scenarios}"
        )

    output_record_count = len(portfolio_pd)
    records_processed = output_record_count

    pd_factor_distribution = (
        portfolio_pd[["Scenario_Name", "PD_Factor"]]
        .value_counts(dropna=False)
        .rename("Record_Count")
        .reset_index()
        .sort_values(by="PD_Factor")
        .reset_index(drop=True)
    )

    print("=" * 80)
    print("PD FACTOR ASSIGNMENT COMPLETED")
    print("=" * 80)
    display(pd_factor_validation)
    print("\nScenario PD Factor Distribution:")
    display(pd_factor_distribution)
    print("Missing PD Factors:", missing_pd_factor)
    print("Invalid PD Factors:", invalid_pd_factor)
    print("Unexpected Scenarios:", unexpected_scenarios)
    print("PD Factor Validation Status: PASS")

    logger.info(
        "Scenario PD Factor assignment completed successfully."
    )
    block_status = "SUCCESS"

except Exception:
    block_status = "FAILED"
    logger.exception("Block 4 failed during PD Factor assignment.")
    raise

finally:
    block_end_time = datetime.now()
    block_duration_seconds = round(
        (block_end_time - block_start_time).total_seconds(), 3
    )

    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```

```python
# =============================================================================
# CELL 06 - BLOCK 5: CALCULATE STRESSED PD AND ASSIGN PD BAND
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(portfolio_pd)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 06: PD STRESS IMPACT")
print("BLOCK 5 : CALCULATE STRESSED PD AND PD BAND")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    portfolio_pd["Stressed_PD"] = (
        portfolio_pd["Base_PD"] * portfolio_pd["PD_Factor"]
    )

    pd_band_conditions = [
        portfolio_pd["Stressed_PD"] <= 0.03,
        portfolio_pd["Stressed_PD"] <= 0.08,
    ]

    pd_band_values = ["Low PD", "Medium PD"]

    portfolio_pd["PD_Band"] = np.select(
        pd_band_conditions,
        pd_band_values,
        default="High PD",
    )

    missing_stressed_pd = int(portfolio_pd["Stressed_PD"].isna().sum())

    invalid_stressed_pd = int(
        (
            (portfolio_pd["Stressed_PD"] < 0)
            | (portfolio_pd["Stressed_PD"] > 1)
        ).sum()
    )

    missing_pd_band = int(portfolio_pd["PD_Band"].isna().sum())

    allowed_pd_bands = ["Low PD", "Medium PD", "High PD"]

    invalid_pd_band = int(
        (~portfolio_pd["PD_Band"].isin(allowed_pd_bands)).sum()
    )

    formula_validation_count = int(
        (
            ~np.isclose(
                portfolio_pd["Stressed_PD"],
                portfolio_pd["Base_PD"] * portfolio_pd["PD_Factor"],
            )
        ).sum()
    )

    if missing_stressed_pd > 0:
        raise ValueError("Missing Stressed_PD values were created.")

    if invalid_stressed_pd > 0:
        raise ValueError(
            "Stressed_PD contains values outside the valid 0-to-1 range."
        )

    if missing_pd_band > 0 or invalid_pd_band > 0:
        raise ValueError("Invalid or missing PD Band values were created.")

    if formula_validation_count > 0:
        raise ValueError("Stressed_PD formula validation failed.")

    output_record_count = len(portfolio_pd)
    records_processed = output_record_count

    pd_band_distribution = pd.crosstab(
        portfolio_pd["Scenario_Name"],
        portfolio_pd["PD_Band"],
        margins=True,
        margins_name="Total",
    )

    print("=" * 80)
    print("STRESSED PD AND PD BAND CALCULATION COMPLETED")
    print("=" * 80)
    print("Missing Stressed PD Values:", missing_stressed_pd)
    print("Invalid Stressed PD Values:", invalid_stressed_pd)
    print("Formula Validation Exceptions:", formula_validation_count)
    print("Missing PD Bands:", missing_pd_band)
    print("Invalid PD Bands:", invalid_pd_band)
    print("Stressed PD Formula Validation: PASS")
    print("\nScenario vs PD Band Distribution:")
    display(pd_band_distribution)
    print("\nSample PD Stress Records:")

    display(
        portfolio_pd[
            [
                "Customer_ID",
                "Loan_ID",
                "Risk_Segment",
                "Scenario_Name",
                "Base_PD",
                "PD_Factor",
                "Stressed_PD",
                "PD_Band",
            ]
        ].head(25)
    )

    logger.info(
        "Stressed PD and PD Band calculations completed successfully."
    )
    block_status = "SUCCESS"

except Exception:
    block_status = "FAILED"
    logger.exception("Block 5 failed during Stressed PD calculation.")
    raise

finally:
    block_end_time = datetime.now()
    block_duration_seconds = round(
        (block_end_time - block_start_time).total_seconds(), 3
    )

    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```

```python
# =============================================================================
# CELL 07 - BLOCK 6: GENERATE SCENARIO-WISE PD SUMMARY
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(portfolio_pd)
output_record_count = 0
records_processed = 0

print("=" * 80)
print("MODULE 06: PD STRESS IMPACT")
print("BLOCK 6 : GENERATE SCENARIO-WISE PD SUMMARY")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    pd_scenario_summary = (
        portfolio_pd.groupby("Scenario_Name", sort=False)
        .agg(
            Record_Count=("Loan_ID", "size"),
            Base_PD_Mean=("Base_PD", "mean"),
            Base_PD_Minimum=("Base_PD", "min"),
            Base_PD_Maximum=("Base_PD", "max"),
            Stressed_PD_Mean=("Stressed_PD", "mean"),
            Stressed_PD_Minimum=("Stressed_PD", "min"),
            Stressed_PD_Maximum=("Stressed_PD", "max"),
        )
        .reset_index()
    )

    numeric_summary_columns = [
        "Base_PD_Mean",
        "Base_PD_Minimum",
        "Base_PD_Maximum",
        "Stressed_PD_Mean",
        "Stressed_PD_Minimum",
        "Stressed_PD_Maximum",
    ]

    pd_scenario_summary[numeric_summary_columns] = (
        pd_scenario_summary[numeric_summary_columns].round(4)
    )

    expected_scenario_counts = {
        "Baseline": 10000,
        "Mild_Stress": 10000,
        "Severe_Stress": 10000,
    }

    pd_scenario_summary["Expected_Record_Count"] = (
        pd_scenario_summary["Scenario_Name"].map(expected_scenario_counts)
    )

    pd_scenario_summary["Status"] = np.where(
        pd_scenario_summary["Record_Count"]
        == pd_scenario_summary["Expected_Record_Count"],
        "PASS",
        "FAIL",
    )

    failed_scenario_summaries = pd_scenario_summary.loc[
        pd_scenario_summary["Status"] != "PASS"
    ]

    if not failed_scenario_summaries.empty:
        raise ValueError("Scenario-wise PD summary validation failed.")

    output_record_count = len(pd_scenario_summary)
    records_processed = input_record_count

    print("=" * 80)
    print("SCENARIO-WISE PD SUMMARY")
    print("=" * 80)
    display(pd_scenario_summary)
    print("Scenario-wise PD Summary Status: PASS")

    logger.info("Scenario-wise PD summary generated successfully.")
    block_status = "SUCCESS"

except Exception:
    block_status = "FAILED"
    logger.exception("Block 6 failed while generating PD summary.")
    raise

finally:
    block_end_time = datetime.now()
    block_duration_seconds = round(
        (block_end_time - block_start_time).total_seconds(), 3
    )

    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)


```

```python
# =============================================================================
# CELL 08 - BLOCK 7: FINAL VALIDATION AND RECONCILIATION
# =============================================================================

block_start_time = datetime.now()
block_status = "STARTED"
input_record_count = len(portfolio_scenario)
output_record_count = len(portfolio_pd)
records_processed = 0

print("=" * 80)
print("MODULE 06: PD STRESS IMPACT")
print("BLOCK 7 : FINAL VALIDATION AND RECONCILIATION")
print("START TIME :", block_start_time.strftime("%Y-%m-%d %H:%M:%S"))
print("=" * 80)

try:
    final_duplicate_keys = int(
        portfolio_pd[["Loan_ID", "Scenario_ID"]].duplicated().sum()
    )

    final_missing_base_pd = int(portfolio_pd["Base_PD"].isna().sum())
    final_missing_pd_factor = int(portfolio_pd["PD_Factor"].isna().sum())
    final_missing_stressed_pd = int(portfolio_pd["Stressed_PD"].isna().sum())
    final_missing_pd_band = int(portfolio_pd["PD_Band"].isna().sum())

    pd_band_total = int(pd_band_distribution.loc["Total", "Total"])

    validation_results = pd.DataFrame({
        "Validation_Check": [
            "Input and Output Record Count",
            "Expected Output Record Count",
            "Duplicate Loan-Scenario Keys",
            "Missing Base PD",
            "Missing PD Factor",
            "Missing Stressed PD",
            "Missing PD Band",
            "Invalid Base PD",
            "Invalid Stressed PD",
            "Formula Validation Exceptions",
            "PD Band Distribution Total",
        ],
        "Expected_Result": [
            input_record_count,
            30000,
            0,
            0,
            0,
            0,
            0,
            0,
            0,
            0,
            30000,
        ],
        "Actual_Result": [
            output_record_count,
            output_record_count,
            final_duplicate_keys,
            final_missing_base_pd,
            final_missing_pd_factor,
            final_missing_stressed_pd,
            final_missing_pd_band,
            invalid_base_pd,
            invalid_stressed_pd,
            formula_validation_count,
            pd_band_total,
        ],
    })

    validation_results["Status"] = np.where(
        validation_results["Expected_Result"].astype(str)
        == validation_results["Actual_Result"].astype(str),
        "PASS",
        "FAIL",
    )

    pd_stress_summary = pd.DataFrame({
        "Metric": [
            "Input Records",
            "Output Records",
            "Input Columns",
            "Output Columns",
            "Low Risk Base PD",
            "Medium Risk Base PD",
            "High Risk Base PD",
            "Other Base PD",
            "Baseline PD Factor",
            "Mild Stress PD Factor",
            "Severe Stress PD Factor",
            "Minimum Stressed PD",
            "Maximum Stressed PD",
            "Low PD Band Records",
            "Medium PD Band Records",
            "High PD Band Records",
            "Processing Timestamp",
        ],
        "Result": [
            input_record_count,
            output_record_count,
            input_column_count,
            len(portfolio_pd.columns),
            0.02,
            0.05,
            0.10,
            0.15,
            1.00,
            1.25,
            1.75,
            float(portfolio_pd["Stressed_PD"].min()),
            float(portfolio_pd["Stressed_PD"].max()),
            int((portfolio_pd["PD_Band"] == "Low PD").sum()),
            int((portfolio_pd["PD_Band"] == "Medium PD").sum()),
            int((portfolio_pd["PD_Band"] == "High PD").sum()),
            datetime.now(),
        ],
    })

    failed_validations = validation_results.loc[
        validation_results["Status"] != "PASS"
    ]

    if not failed_validations.empty:
        raise ValueError("One or more Module 06 validation checks failed.")

    records_processed = output_record_count

    print("=" * 80)
    print("FINAL PD STRESS IMPACT VALIDATION")
    print("=" * 80)
    display(validation_results)
    print("\nPD Stress Module Summary:")
    display(pd_stress_summary)
    print("Overall Module 06 Validation Status: PASS")

    logger.info("All Module 06 validation checks passed.")
    block_status = "SUCCESS"

except Exception:
    block_status = "FAILED"
    logger.exception("Block 7 failed during final validation.")
    raise

finally:
    block_end_time = datetime.now()
    block_duration_seconds = round(
        (block_end_time - block_start_time).total_seconds(), 3
    )

    print("=" * 80)
    print("BLOCK STATUS       :", block_status)
    print("INPUT RECORDS      :", input_record_count)
    print("OUTPUT RECORDS     :", output_record_count)
    print("RECORDS PROCESSED  :", records_processed)
    print("END TIME           :", block_end_time.strftime("%Y-%m-%d %H:%M:%S"))
    print("DURATION (SECONDS) :", block_duration_seconds)
    print("=" * 80)

```
