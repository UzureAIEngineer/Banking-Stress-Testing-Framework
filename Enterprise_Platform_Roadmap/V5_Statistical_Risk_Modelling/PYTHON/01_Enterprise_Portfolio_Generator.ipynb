### Module 01 - Enterprise Portfolio Validation

Dataset Summary

- Retail Credit Portfolio: 100,000 Records
- Wholesale Credit Portfolio: 100,000 Records
- Total Portfolio: 200,000 Records

Purpose

- Read SAS-generated Enterprise Portfolio Dataset
- Perform Data Validation
- Verify Portfolio Distribution
- Perform Descriptive Analysis
- Support SAS to Python Reconciliation

```python
# ==========================================================
# IMPORT LIBRARIES
# ==========================================================

import pandas as pd
import numpy as np
import time
from datetime import datetime

# ==========================================================
# EXECUTION START
# ==========================================================

start_time = time.time()

print("=" * 60)
print("V5 - ENTERPRISE PORTFOLIO VALIDATION")
print("=" * 60)
print(f"Execution Started : {datetime.now()}")
print("=" * 60)
```

```python
# ==========================================================
# READ SAS GENERATED DATASET
# ==========================================================

file_path = "../DATA/enterprise_portfolio_200k.csv"

df = pd.read_csv(file_path)

print("Dataset Loaded Successfully")
print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]:,}")
```

```python
# ==========================================================
# COLUMN VALIDATION
# ==========================================================

print("=" * 60)
print("COLUMN VALIDATION")
print("=" * 60)

for col in df.columns:
    print(col)
```

```python
# ==========================================================
# PORTFOLIO DISTRIBUTION
# ==========================================================

portfolio_dist = (
    df["PORTFOLIO_TYPE"]
    .value_counts()
)

print(portfolio_dist)
```

```python
# ==========================================================
# PRODUCT DISTRIBUTION
# ==========================================================

product_dist = (
    df["PRODUCT_TYPE"]
    .value_counts()
    .sort_index()
)

print(product_dist)
```

```python
# ==========================================================
# MISSING VALUE VALIDATION
# ==========================================================

missing_values = df.isnull().sum()

print(
    missing_values[
        missing_values > 0
    ]
)
```

```python
# ==========================================================
# DESCRIPTIVE STATISTICS
# ==========================================================

numeric_cols = [
    "CREDIT_SCORE",
    "ANNUAL_INCOME",
    "LOAN_AMOUNT",
    "OUTSTANDING_BALANCE",
    "LTV",
    "DTI",
    "PD",
    "LGD",
    "EAD"
]

df[numeric_cols].describe()
```

```python
# ==========================================================
# SAMPLE PORTFOLIO RECORDS
# ==========================================================

df.head(10)
```

```python
# ==========================================================
# EXECUTION SUMMARY
# ==========================================================

end_time = time.time()

duration = round(
    end_time - start_time,
    2
)

print("=" * 60)
print("VALIDATION COMPLETED")
print("=" * 60)

print(f"Total Records      : {len(df):,}")
print(f"Execution Duration : {duration} seconds")

print("=" * 60)
```
